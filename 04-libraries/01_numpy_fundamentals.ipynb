## 📖 Introduction

### What is NumPy?

NumPy is the fundamental package for **numerical computing** in Python. It
provides the `ndarray` type for efficient array operations, vectorized
computations, broadcasting, and many mathematical functions implemented in C.

### Key Features:
- Efficient multidimensional arrays (`ndarray`)
- Fast element-wise operations and vectorization
- Broadcasting rules for combining arrays of different shapes
- Linear algebra, statistics, and random number utilities

## 🔑 Core Concept 1: Arrays and Basic Operations\n\nThe core building block of NumPy is the `ndarray`. You create arrays from Python\nlists and then perform fast vectorized operations on them.\n\n### Example:


````python
import numpy as np\n\n# Create 1D and 2D arrays\na = np.array([1, 2, 3])\nb = np.array([[1, 2, 3], [4, 5, 6]])\nprint("a =", a)\nprint("b =", b)\nprint("shape of b =", b.shape)\n\n# Vectorized operations\nprint("a + 10 =", a + 10)\nprint("a * 2  =", a * 2)\nprint("a ** 2 =", a ** 2)\n\n# Element-wise operations between arrays of the same shape\nc = np.array([10, 20, 30])\nprint("a + c =", a + c)\n
````

## 🔑 Core Concept 2: Indexing, Slicing, and Boolean Masks\n\nIndexing and slicing let you select subsets of data efficiently. Boolean masks\nare especially powerful for filtering arrays based on conditions.\n\n### Example:


````python
import numpy as np\n\narr = np.arange(10)  # [0, 1, ..., 9]\nprint("arr =", arr)\nprint("arr[0:5] =", arr[0:5])  # slicing\nprint("arr[::2]  =", arr[::2])  # step slicing\n\n# Boolean masking\nmask = arr % 2 == 0  # even numbers\nprint("mask =", mask)\nprint("arr[mask] =", arr[mask])\n
````

## 🔑 Core Concept 3: Broadcasting\n\nBroadcasting describes how NumPy automatically expands arrays of different\nshapes to perform element-wise operations without explicit loops.\n\n### Example:


````python
import numpy as np\n\n# Add a 1D array to each row of a 2D array\nX = np.array([[1, 2, 3], [4, 5, 6]])\nv = np.array([10, 20, 30])\nprint("X =
", X)\nprint("v =", v)\nprint("X + v =
", X + v)  # v is broadcast across rows\n\n# Broadcasting with reshaping\ncol = np.array([1, 2]).reshape(-1, 1)\nprint("X * col =
", X * col)\n
````

## ✅ Best Practices

When working with NumPy, keep these best practices in mind:

1. **Use vectorization** instead of Python loops whenever possible.
2. **Leverage broadcasting** instead of manual tiling/repetition of arrays.
3. **Be explicit with dtypes** when precision matters (e.g., float32 vs float64).
4. **Avoid unnecessary copies** by understanding views vs copies when slicing.
5. **Use `np.random.Generator`** for reproducible random numbers.

### Common Pitfalls to Avoid:
- Assuming slicing always creates a copy (it often returns a view).
- Mixing Python lists and NumPy arrays in arithmetic expressions.
- Forgetting to handle NaNs or invalid values before aggregations.
