# Lesson 2: Variables and Data Types

## What are Variables?

Variables are containers for storing data values. Think of them as labeled boxes where you can store information.

### Variable Naming Rules:
- Must start with a letter or underscore
- Can contain letters, numbers, and underscores
- Case-sensitive (`name` and `Name` are different)
- Cannot use Python keywords (like `print`, `if`, `for`)

---

## Creating Variables

```python
# Creating variables
name = "Alice"
age = 25
height = 5.6
is_student = True

print("Name:", name)
print("Age:", age)
print("Height:", height)
print("Is student:", is_student)
```

## Python Data Types

Python has several built-in data types:

| Data Type | Description | Example |
|-----------|-------------|----------|
| `int` | Integer numbers | `42`, `-10`, `0` |
| `float` | Decimal numbers | `3.14`, `-0.5`, `2.0` |
| `str` | Text/strings | `"Hello"`, `'Python'` |
| `bool` | True/False values | `True`, `False` |

---

## Working with Integers (int)

```python
# Integer examples
score = 100
temperature = -5
year = 2024

print("Score:", score)
print("Type:", type(score))
```

## Working with Floats (float)

```python
# Float examples
pi = 3.14159
price = 19.99
weight = 68.5

print("Pi:", pi)
print("Type:", type(pi))
```

## Working with Strings (str)

```python
# String examples
greeting = "Hello, World!"
language = 'Python'
message = """This is a
multi-line string"""

print(greeting)
print("Type:", type(greeting))
```

## Working with Booleans (bool)

```python
# Boolean examples
is_sunny = True
is_raining = False

print("Is sunny:", is_sunny)
print("Type:", type(is_sunny))
```

## Type Conversion

You can convert between data types:

```python
# Type conversion examples
num_str = "100"
num_int = int(num_str)  # String to integer
print("Converted:", num_int, "Type:", type(num_int))

num_float = float(num_str)  # String to float
print("Converted:", num_float, "Type:", type(num_float))

age = 25
age_str = str(age)  # Integer to string
print("Converted:", age_str, "Type:", type(age_str))
```

## Variable Reassignment

```python
# Variables can be reassigned
x = 10
print("Initial value:", x)

x = 20
print("New value:", x)

x = "Now I'm a string!"
print("Changed type:", x)
```

---
## 💡 Key Takeaways

- Variables store data values
- Python has main data types: int, float, str, bool
- Use `type()` to check data type
- Variables can be reassigned to different values and types
- Type conversion: `int()`, `float()`, `str()`
