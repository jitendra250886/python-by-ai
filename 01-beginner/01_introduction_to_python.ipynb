### What is Python?
Python is a high-level, interpreted programming language known for:
- **Easy to learn**: Clean and readable syntax
- **Versatile**: Web development, data science, AI, automation, and more
- **Large community**: Extensive libraries and support
- **Cross-platform**: Works on Windows, Mac, and Linux

### Why Learn Python?

2. High demand in job market
3. Powerful libraries for any task

## Basic Print Statements

The `print()` function displays output to the screen:

````python
# Printing different types of content
print("Welcome to Python!")
print(42)
print(3.14)
print(True)
````
